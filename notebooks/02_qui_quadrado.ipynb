Este notebook realiza **testes qui-quadrado** para verificar independência entre variáveis categóricas.

### Objetivo:
Identificar se existe associação significativa entre pares de variáveis.

### Metodologia:
- Teste Qui-Quadrado de Pearson (χ²)
- Nível de significância: α = 0.05
- H0: As variáveis são independentes
- H1: As variáveis NÃO são independentes

## Testes Qui-Quadrado

Vamos testar associações entre variáveis relevantes para a curva de aprendizado.

### Teste 1: Porte da Empresa × Principal Desafio

```python
resultado1 = utils.teste_qui_quadrado(
    df,
    'q6_porte_empresa',
    'q18_principal_desafio',
    "Porte da Empresa × Principal Desafio"
)
```

### Teste 2: Empresa Ofereceu Treinamento × Tempo Individual

```python
resultado2 = utils.teste_qui_quadrado(
    df,
    'q20_empresa_ofereceu_treinamento',
    'q24_tempo_individual',
    "Empresa Ofereceu Treinamento × Tempo Individual"
)
```

### Teste 3: Realizou POCs × Tempo da Equipe

```python
resultado3 = utils.teste_qui_quadrado(
    df,
    'q23_realizou_pocs',
    'q25_tempo_equipe',
    "Realizou POCs × Tempo da Equipe"
)
```

### Teste 4: Fonte Mais Útil × Tempo Individual

```python
resultado4 = utils.teste_qui_quadrado(
    df,
    'q22_fonte_mais_util',
    'q24_tempo_individual',
    "Fonte Mais Útil × Tempo Individual"
)
```

### Teste 5: Cargo × Tempo com Containers

---
## Resumo dos Resultados

```python
utils.titulo_secao("RESUMO DOS TESTES QUI-QUADRADO")

resultados = [
    ("Porte da Empresa × Principal Desafio", resultado1),
    ("Empresa Ofereceu Treinamento × Tempo Individual", resultado2),
    ("Realizou POCs × Tempo da Equipe", resultado3),
    ("Fonte Mais Útil × Tempo Individual", resultado4),
    ("Cargo × Tempo com Containers", resultado5),
    ("Setor × Tecnologia de Orquestração", resultado6)
]

alpha = 0.05
print(f"\nNível de significância: α = {alpha}\n")

for nome, res in resultados:
    significativo = "SIM" if res['p_value'] < alpha else "NÃO"
    print(f"{nome}")
    print(f"   χ² = {res['chi2']:.4f}, p = {res['p_value']:.4f}")
    print(f"   Associação significativa: {significativo}")
    print()
```

## Interpretação

Os testes qui-quadrado identificaram quais pares de variáveis apresentam **associação significativa** (p < 0.05).
