# Teste Qui-Quadrado de Independência

Associação entre os desafios percebidos (**Q12–Q17**, Likert) e a velocidade da curva de aprendizado.

**Bateria definida a priori:** cada desafio é dicotomizado (alto = 4-5) e cruzado com os dois desfechos — curva da equipe (**Q25**, até 3 meses) e curva individual (**Q24**, até 6 meses). Resultam 12 tabelas 2×2.

**Critérios:** α = 0,05 · premissa das frequências esperadas verificada (máx. 20% das células com esperado < 5, nenhuma < 1) · Fisher exato como p de referência em 2×2.

---
## Configuração

```python
import sys; sys.path.append('../src')
import pandas as pd, utils

utils.configurar_visualizacoes()

# Reaproveita o df do master, ou carrega e trata se rodando isolado
df = utils.garantir_preparado(globals().get('df'))
print(f"{len(df)} respostas × {len(df.columns)} variáveis")
```

---
## Bateria de testes

Os 12 pares são montados a partir de `utils.DESAFIOS_LIKERT`, garantindo que a bateria seja explícita e reprodutível.

### Definição dos pares e execução

```python
# Bloco principal: desafios x curva de aprendizado da EQUIPE
PARES_EQUIPE = [
    (col + '_alto', 'curva_rapida_equipe', f"{rotulo} × curva rápida da equipe")
    for rotulo, col in utils.DESAFIOS_LIKERT.items()
]

# Bloco secundário: desafios x curva de aprendizado INDIVIDUAL
PARES_INDIVIDUAL = [
    (col + '_alto', 'curva_rapida_individual', f"{rotulo} × curva rápida individual")
    for rotulo, col in utils.DESAFIOS_LIKERT.items()
]

PARES = PARES_EQUIPE + PARES_INDIVIDUAL
print(f"Bateria definida: {len(PARES)} testes\n")

bateria = utils.bateria_qui_quadrado(df, PARES, alpha=0.05, exibir_tabelas=True)

# Publicado para o notebook 03 (ANACOR) consumir
resultados_qui = bateria['resultados']
resumo_qui = bateria['resumo']
```

## Interpretação

**Significativo:** `sim` (p < α com premissa atendida) = achado · `não` = resultado nulo · `não interpretável` = premissa violada.

**V de Cramér** mede o tamanho do efeito: 0,1 fraco · 0,3 médio · 0,5 forte. Não significativo com efeito médio indica falta de poder, não ausência de associação.

Só os pares `sim` são habilitados à ANACOR.
